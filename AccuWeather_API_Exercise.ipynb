## Step 4 — Build the API URL

The AccuWeather Current Conditions endpoint follows this pattern:

```
http://dataservice.accuweather.com/currentconditions/v1/{LOCATION_ID}
```

We also pass two **parameters**:
- `apikey` — your credentials
- `details` — set to `true` so we get extra detail like wind speed and humidity


````python
BASE_URL = "http://dataservice.accuweather.com/currentconditions/v1"
url = f"{BASE_URL}/{LOCATION_ID}"

params = {
    "apikey": API_KEY,
    "details": "true"
}

print(f"✅ URL ready: {url}")
print(f"✅ Parameters: {list(params.keys())}")
````

## Step 5 — Send the GET Request

This is where we actually call the API.

`requests.get()` sends a GET request to the URL with our parameters.  
The server receives it, checks our API key, finds the weather data, and sends it back.

We then check the **status code**:
- `200` = success ✅
- `401` = wrong API key ❌
- `404` = location not found ❌
- `500` = server error ❌


````python
response = requests.get(url, params=params)

print(f"Status Code: {response.status_code}")

if response.status_code == 200:
    print("✅ Request successful! Data received.")
elif response.status_code == 401:
    print("❌ Unauthorized — check your API key")
elif response.status_code == 404:
    print("❌ Not Found — check your location ID")
else:
    print(f"❌ Something went wrong: {response.status_code}")
````

## Step 6 — View the Raw JSON Response

The API returns data in **JSON format** — a structured text format that looks like a Python dictionary.

Let's print the full response so you can see everything AccuWeather sent back.


````python
if response.status_code == 200:
    raw_data = response.json()
    print(json.dumps(raw_data, indent=2))
else:
    print("Cannot display data — request was not successful.")
````

## Step 7 — Extract Specific Fields

Now let's pull out only the fields we care about.

The response is a **list** — `raw_data[0]` gets the first (and only) item which contains the current conditions.


````python
if response.status_code == 200:
    weather = raw_data[0]

    # Extract key fields
    temperature_c   = weather["Temperature"]["Metric"]["Value"]
    temperature_f   = weather["Temperature"]["Imperial"]["Value"]
    condition       = weather["WeatherText"]
    is_daytime      = weather["IsDayTime"]
    humidity        = weather.get("RelativeHumidity", "N/A")
    wind_speed_kmh  = weather["Wind"]["Speed"]["Metric"]["Value"]
    wind_direction  = weather["Wind"]["Direction"]["Localized"]
    uv_index        = weather.get("UVIndex", "N/A")
    uv_text         = weather.get("UVIndexText", "N/A")
    fetched_at      = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    print("=" * 45)
    print("        CURRENT WEATHER CONDITIONS")
    print("=" * 45)
    print(f"  📍 Location ID   : {LOCATION_ID}")
    print(f"  🕐 Fetched At    : {fetched_at}")
    print(f"  🌡️  Temperature   : {temperature_c}°C  /  {temperature_f}°F")
    print(f"  🌤️  Condition     : {condition}")
    print(f"  ☀️  Daytime       : {'Yes' if is_daytime else 'No'}")
    print(f"  💧 Humidity      : {humidity}%")
    print(f"  🌬️  Wind          : {wind_speed_kmh} km/h {wind_direction}")
    print(f"  🔆 UV Index      : {uv_index} ({uv_text})")
    print("=" * 45)
````

## Step 8 — Save to a CSV File

Now let's write what we fetched into a CSV file so we can keep a log over time.

Every time you run this cell, it **adds a new row** to `weather_log.csv` with the current reading and timestamp.

> This is exactly how real data pipelines work — fetch → extract → store


````python
import csv
import os

csv_file = "weather_log.csv"

# Write header only if file does not exist yet
file_exists = os.path.exists(csv_file)

with open(csv_file, "a", newline="") as f:
    writer = csv.writer(f)

    if not file_exists:
        writer.writerow([
            "Timestamp", "Temperature_C", "Temperature_F",
            "Condition", "Is_Daytime", "Humidity_%",
            "Wind_Speed_kmh", "Wind_Direction", "UV_Index"
        ])

    writer.writerow([
        fetched_at, temperature_c, temperature_f,
        condition, is_daytime, humidity,
        wind_speed_kmh, wind_direction, uv_index
    ])

print(f"✅ Data saved to {csv_file}")
print(f"   Row added at: {fetched_at}")
````

## Step 9 — Read Back the CSV File

Let's confirm the data was saved correctly by reading the CSV back and displaying it as a table.


````python
import pandas as pd

df = pd.read_csv(csv_file)
print(f"✅ {len(df)} row(s) in weather_log.csv\n")
display(df)
````

## 🏆 Challenge — Auto-fetch Every 2 Hours

This is the final challenge. The loop below will:
1. Call the AccuWeather API
2. Extract the weather fields
3. Save them to the CSV
4. Wait 2 hours
5. Repeat

> ⚠️ **To stop the loop**, click the ■ Stop button in the toolbar above, or press `I` twice (interrupt kernel).  
> Run this only when you want to leave it running in the background.


````python
import time

print("🔄 Auto-fetch started. Press the Stop button to end.\n")

cycle = 1
while True:
    print(f"--- Cycle {cycle} | {datetime.now().strftime('%Y-%m-%d %H:%M:%S')} ---")

    resp = requests.get(url, params=params)

    if resp.status_code == 200:
        w = resp.json()[0]
        row_time     = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
        t_c          = w["Temperature"]["Metric"]["Value"]
        t_f          = w["Temperature"]["Imperial"]["Value"]
        cond         = w["WeatherText"]
        day          = w["IsDayTime"]
        hum          = w.get("RelativeHumidity", "N/A")
        wind_s       = w["Wind"]["Speed"]["Metric"]["Value"]
        wind_d       = w["Wind"]["Direction"]["Localized"]
        uv           = w.get("UVIndex", "N/A")

        with open(csv_file, "a", newline="") as f:
            writer = csv.writer(f)
            writer.writerow([row_time, t_c, t_f, cond, day, hum, wind_s, wind_d, uv])

        print(f"  ✅ Saved: {t_c}°C, {cond}, Humidity: {hum}%")
    else:
        print(f"  ❌ Request failed: {resp.status_code}")

    cycle += 1
    print(f"  ⏳ Waiting 2 hours until next fetch...\n")
    time.sleep(7200)
````
